Which category generates the highest sales?

```python
data=pd.read_csv('./Data.csv',encoding='latin 1')
```

Which category generates the highest profit?

```python
data.groupby(data['Category'])['Profit'].sum().sort_values(ascending=False).head(1)
```

Which region generates the highest sales?

```python
data.groupby(data['Region'])['Sales'].sum().sort_values(ascending=False).head(1)
```

Which region has the lowest profit?

```python
data.groupby(data['Region'])['Profit'].sum().sort_values(ascending=True).head(1)
```

Which sub-category has the highest sales?

Which sub-category has the highest profit?

```python
data.groupby(data['Sub-Category'])['Profit'].sum().sort_values(ascending=False).head(1)
```

Find the top 5 states based on total sales.

```python
data.groupby(data['State'])['Profit'].sum().sort_values(ascending=False).head(5)
```

```python
data.groupby(data['Customer Name'])['Quantity'].sum().sort_values(ascending=False).head(10)
```

```python
data.groupby(data['Category'])['Sales'].mean().sort_values(ascending=False).head(1)
```
